## A3. El bucle `for`

### Qué vamos a lograr
Repetir una operación para cada posición de una lista, sin escribirla una por una.

### La idea
Una neurona con $784$ entradas no se puede calcular escribiendo $784$ términos. Una **sumatoria** $\sum_{i=1}^{n}$ resuelve eso en matemáticas; el **bucle** `for` lo resuelve en Python:

```
for i in range(3):
    ...instrucciones...
```

- `range(3)` genera los números $0,1,2$ (empieza en 0 y se detiene **antes** del 3).
- En cada vuelta, `i` toma uno de esos valores.
- Lo que se repite va con **sangría** (espacios al inicio). En Python la sangría es obligatoria: es la forma de indicar qué está dentro del bucle.

### Ejemplo resuelto (solo ejecútalo)
Observa cómo `i` cambia y cómo `x[i]` recorre la lista.

````
x = [2, 1, 3]

for i in range(3):
    print("i =", i, "   x[i] =", x[i])
````

### Resultado esperado
Tres líneas: `i = 0 x[i] = 2`, `i = 1 x[i] = 1`, `i = 2 x[i] = 3`.

## A4. El acumulador (sumar muchas cosas)

### Qué vamos a lograr
Sumar todos los elementos de una lista usando un bucle.

### La idea
Una suma de muchos términos se construye por **sumas parciales**:

$$S_0=0,\qquad S_k = S_{k-1} + a_k$$

En Python esto se escribe con una variable que **va creciendo**:

```
total = 0
total = total + 5
```

La segunda línea no es una ecuación: dice "toma lo que había en `total`, súmale 5 y guárdalo de nuevo en `total`". Dentro de un bucle, esa línea se ejecuta una vez por elemento.

### Ejemplo resuelto (solo ejecútalo)
Sumamos $1+2+3$.

````
numeros = [1, 2, 3]

total = 0                       # S_0 = 0
for i in range(3):
    total = total + numeros[i]  # S_k = S_(k-1) + a_k
    print("después de la vuelta", i, "el total es", total)
````

# Parte B: la neurona

## La idea completa, antes de programar

### Una neurona es una votación ponderada
Imagina que quieres decidir si una foto es de un gato. Tienes varios datos de entrada $x_1,\dots,x_n$ (por ejemplo, el brillo de cada píxel). Cada dato "vota" a favor o en contra, pero **no todos los votos valen lo mismo**. El peso $w_i$ dice cuánto vale el voto del dato $i$:

- $w_i>0$: ese dato empuja el resultado hacia arriba.
- $w_i<0$: lo empuja hacia abajo.
- $w_i\approx 0$: casi no influye.

Se suman todos los votos ponderados y se añade un número fijo $b$, el **sesgo** (*bias*):

$$y=\sum_{i=1}^{n} w_i x_i + b = w^{\top}x + b$$

### ¿Para qué sirve el sesgo $b$?
Sin $b$, la neurona valdría siempre $0$ cuando todas las entradas son $0$. Geométricamente, la ecuación $w^{\top}x = 0$ describe un hiperplano que **pasa por el origen**. Con $b$, el hiperplano $w^{\top}x+b=0$ puede estar en cualquier posición: $b$ lo desplaza. Es el término independiente de una ecuación lineal, y es lo que permite que la neurona se active "más fácil" o "más difícil".

### Qué significa $y$
$y$ es un número que mide **cuánto se parece el dato a lo que la neurona busca**. Para clasificar, se mira su signo: $y>0$ de un lado del hiperplano, $y<0$ del otro. De ahí viene que una neurona sea un **clasificador lineal**.

## B4. El producto punto

### Qué vamos a lograr
Calcular la neurona completa en **una sola línea**.

### La idea
La suma de los productos elemento a elemento es exactamente el **producto punto** (producto interno) que ya conoces:

$$w^{\top}x = \sum_{i=1}^{n} w_i x_i$$

NumPy lo calcula con `np.dot(a, b)`: recibe **dos** vectores separados por coma. Entonces la neurona queda

$$y = w^{\top}x + b \;\;\longrightarrow\;\; \texttt{np.dot(w, x) + b}$$

### Ejemplo resuelto (solo ejecútalo)

````
x = np.array([1, 2])
w = np.array([3, 4])
b = 5

y = np.dot(w, x) + b
print(y)       # 16
````

# Parte C: la capa

## La idea completa, antes de programar

### Varias neuronas mirando lo mismo
Una sola neurona produce **un** número. Una **capa** tiene $m$ neuronas que reciben **las mismas $n$ entradas**, pero cada una con **sus propios** pesos y su propio sesgo. Así, cada neurona busca un patrón distinto en los mismos datos, y la capa produce **$m$ números**.

### La matriz $W$
Los pesos de las $m$ neuronas se guardan en una matriz $W\in\mathbb{R}^{m\times n}$ donde **la fila $j$ contiene los pesos de la neurona $j$**:

$$W=\begin{pmatrix} W_{11}&W_{12}&\cdots&W_{1n}\\ W_{21}&W_{22}&\cdots&W_{2n}\\ \vdots&&&\vdots\\ W_{m1}&W_{m2}&\cdots&W_{mn}\end{pmatrix}$$

Los sesgos van en un vector $b\in\mathbb{R}^m$ (uno por neurona). La salida de la neurona $j$ es

$$y_j=\sum_{i=1}^{n} W_{ji}\,x_i + b_j$$

### Es una transformación afín
Si recuerdas Álgebra Lineal, $x\mapsto Wx$ es una **transformación lineal** de $\mathbb{R}^n$ en $\mathbb{R}^m$, y sumar $b$ es una **traslación**. Una transformación lineal más una traslación se llama **afín**. Todas las capas densas de las redes neuronales son transformaciones afines.

### Tabla de dimensiones
| Objeto | Qué es | Forma |
|---|---|---|
| $x$ | entradas | $n$ |
| $W$ | pesos | $m\times n$ |
| $b$ | sesgos | $m$ |
| $y$ | salidas | $m$ |

Regla para no perderte: **las filas de $W$ son las neuronas, las columnas son las entradas**.

# Parte D: muchos datos a la vez

## La idea completa, antes de programar

### Por qué un lote
En la práctica una red no procesa un dato a la vez, sino **un lote** (*batch*) de $N$ ejemplos juntos: es mucho más rápido y es lo que usará el entrenamiento. Guardamos el lote en una matriz $X\in\mathbb{R}^{N\times n}$ donde **cada fila es un ejemplo**.

### Por qué aparece la transpuesta
Para **un** ejemplo, la salida es $y=Wx+b$ con $x$ columna. Pero en el lote cada ejemplo $x_k$ es una **fila** de $X$. Transpongamos la fórmula (recordando que $(AB)^{\top}=B^{\top}A^{\top}$):

$$y_k^{\top} = (Wx_k+b)^{\top} = x_k^{\top}\,W^{\top} + b^{\top}$$

Es decir, la salida del ejemplo $k$ (como fila) es la fila $x_k^\top$ multiplicada por $W^\top$, más $b^\top$. Apilando las $N$ filas:

$$Y = XW^{\top} + \mathbf{1}\,b^{\top}$$

donde $\mathbf{1}$ es un vector de unos que repite el sesgo en cada fila. NumPy hace esa repetición sola (*broadcasting*), así que se escribe `X @ W.T + b`.

### Comprobación de formas
$X$ es $N\times n$, $W^{\top}$ es $n\times m$, luego $XW^{\top}$ es $N\times m$: **una fila por ejemplo, una columna por neurona**. Las dimensiones internas ($n$ con $n$) deben coincidir; si no, NumPy da error.

# Parte E: resumen

## Lo que lograste
1. Una **neurona** es $y=w^{\top}x+b$: una votación ponderada.
2. Una **capa** apila $m$ neuronas en una matriz $W$ (cada fila, una neurona): $y=Wx+b$, una transformación afín.
3. Un **lote** de $N$ ejemplos se procesa de una vez con $Y=XW^{\top}+\mathbf{1}b^{\top}$.
4. `.shape` te dice si las dimensiones encajan.

## Qué falta (próximas lecciones)
Si apilas dos capas afines sin nada entre ellas, el resultado sigue siendo **una sola** transformación afín: no ganas poder. Por eso en la siguiente unidad añadimos una **función de activación** no lineal entre capas. Y luego veremos cómo se ajustan $W$ y $b$ (el entrenamiento).

Más adelante empaquetaremos lo anterior en una *clase* para poder apilar capas. Por ahora basta con que veas que es **el mismo cálculo**, con $W$ y $b$ guardados dentro del objeto.

````
class CapaDensa:
    def __init__(self, n_entradas, n_neuronas):
        self.W = 0.01 * np.random.randn(n_neuronas, n_entradas)   # (neuronas, entradas)
        self.b = np.zeros(n_neuronas)                              # un sesgo por neurona

    def forward(self, X):
        return X @ self.W.T + self.b                               # el mismo cálculo de D2

capa = CapaDensa(3, 4)
print(capa.forward(X).shape)    # (2, 4)
````
